##  2) Crear el chatbot utilizando TFIDF y similitud del coseno. (1 punto)

**Uso de IA:** Se utilizó ChatGPT como apoyo para comprender el funcionamiento

la función de recuperación. El código fue probado y analizado durante el

### Preguntas de Prueba

```python
preguntas_prueba = [

    # 5 consultas con vocabulario parecido al dataset
    "¿Qué significa población en estadística?",
    "¿Cómo se obtiene la frecuencia relativa?",
    "¿Cómo puedo calcular la media aritmética?",
    "¿Qué significa desviación estándar?",
    "¿Para qué se utiliza un gráfico de barras?",

    # 10 reformulaciones naturales
    "¿Qué llamamos muestra en un estudio estadístico?",
    "¿Cómo puedo saber si una variable es cualitativa o cuantitativa?",
    "¿En qué se distingue una variable discreta de una continua?",
    "¿Qué indica la frecuencia absoluta de un dato?",
    "¿Para qué sirve organizar los datos en una tabla de frecuencias?",
    "¿Cómo saco el promedio de un conjunto de valores?",
    "¿Qué representa la mediana de un conjunto de datos?",
    "¿Cómo sé cuál es la moda de un conjunto de datos?",
    "¿En qué casos conviene usar la mediana y no la media?",
    "¿Qué diferencia hay entre un histograma y un gráfico de barras?"
]
```

```python
# Crear el vectorizador
vectorizer = TfidfVectorizer(
    lowercase=True,
    strip_accents="unicode"
)

# Ajustar y transformar las preguntas del dataset
tfidf_matrix = vectorizer.fit_transform(preguntas)

# Función para encontrar la mejor coincidencia
def responder_pregunta(consulta_usuario):

    # Transformar la consulta usando el mismo vectorizador
    consulta_vec = vectorizer.transform([consulta_usuario])

    # Calcular la similitud entre la consulta y todas las preguntas
    similitudes = cosine_similarity(
        consulta_vec,
        tfidf_matrix
    ).flatten()

    # Buscar el índice de la pregunta con mayor similitud
    indice_mejor_coincidencia = similitudes.argmax()

    # Guardar los resultados
    respuesta = respuestas[indice_mejor_coincidencia]
    pregunta_encontrada = preguntas[indice_mejor_coincidencia]
    similitud = similitudes[indice_mejor_coincidencia]

    return respuesta, pregunta_encontrada, similitud
```

```python
for consulta in preguntas_prueba:

    respuesta, pregunta_encontrada, similitud = responder_pregunta(consulta)

    print("Consulta:", consulta)
    print("Pregunta encontrada:", pregunta_encontrada)
    print("Similitud:", round(similitud, 3))
    print("Respuesta:", respuesta)
    print("-" * 100)
```

Sin embargo, presentó dificultades ante algunas reformulaciones que utilizaban palabras diferentes para expresar un mismo concepto. Por ejemplo, la consulta "¿Cómo saco el promedio de un conjunto de valores?" no fue relacionada con "media aritmética".

```python
vectores_preguntas = np.array(vectores_preguntas)
vectores_preguntas.shape
```

```python
def responder_embeddings(consulta_usuario):

    # Convertir la consulta del usuario en un embedding
    consulta_vec = nlp(consulta_usuario).vector

    # Comparar la consulta con los vectores de las 40 preguntas
    similitudes = cosine_similarity(
        [consulta_vec],
        vectores_preguntas
    ).flatten()

    # Buscar el índice de la mayor similitud
    indice_mejor_coincidencia = similitudes.argmax()

    # Recuperar la pregunta, respuesta y similitud
    respuesta = respuestas[indice_mejor_coincidencia]
    pregunta_encontrada = preguntas[indice_mejor_coincidencia]
    similitud = similitudes[indice_mejor_coincidencia]

    return respuesta, pregunta_encontrada, similitud
```

```python
for consulta in preguntas_prueba:

    respuesta, pregunta_encontrada, similitud = responder_embeddings(consulta)

    print("Consulta:", consulta)
    print("Pregunta encontrada:", pregunta_encontrada)
    print("Similitud:", round(similitud, 3))
    print("Respuesta:", respuesta)
    print("-" * 80)
```

```python
def obtener_vector_sin_stopwords(texto):

    doc = nlp(texto)

    vectores = [
        token.vector
        for token in doc
        if not token.is_stop
        and not token.is_punct
        and not token.is_space
        and token.has_vector
    ]

    if len(vectores) > 0:
        return np.mean(vectores, axis=0)

    return np.zeros(nlp.vocab.vectors_length)
```

```python
vectores_preguntas_sin_stopwords = []

for pregunta in preguntas:
    vector = obtener_vector_sin_stopwords(pregunta)
    vectores_preguntas_sin_stopwords.append(vector)

vectores_preguntas_sin_stopwords = np.array(
    vectores_preguntas_sin_stopwords
)
```

```python
def responder_embeddings_sin_stopwords(consulta_usuario):

    # Convertir la consulta usando el mismo procesamiento
    consulta_vec = obtener_vector_sin_stopwords(consulta_usuario)

    # Comparar con las 40 preguntas
    similitudes = cosine_similarity(
        [consulta_vec],
        vectores_preguntas_sin_stopwords
    ).flatten()

    # Buscar la mayor similitud
    indice_mejor_coincidencia = similitudes.argmax()

    # Recuperar resultados
    respuesta = respuestas[indice_mejor_coincidencia]
    pregunta_encontrada = preguntas[indice_mejor_coincidencia]
    similitud = similitudes[indice_mejor_coincidencia]

    return respuesta, pregunta_encontrada, similitud
```

```python
for consulta in preguntas_prueba:

    respuesta, pregunta_encontrada, similitud = responder_embeddings_sin_stopwords(
        consulta
    )

    print("Consulta:", consulta)
    print("Pregunta encontrada:", pregunta_encontrada)
    print("Similitud:", round(similitud, 3))
    print("Respuesta:", respuesta)
    print("-" * 80)
```

### Posibles mejoras

En **TF-IDF** se podría incorporar lematización para tratar de la misma manera distintas formas de una palabra. Por ejemplo, "calcula", "calcular" y "calculando" podrían reducirse al lema "calcular". También podrían utilizarse bigramas para considerar combinaciones de palabras y no solamente como términos separados.

En **spaCy**, la eliminación de stopwords puede hacer perder información importante. Por ejemplo, eliminar la palabra "no" podría modificar completamente la intención de una consulta. Además, en esta implementación el vector de cada pregunta se obtuvo calculando el promedio de los embeddings de sus palabras. Al utilizar un promedio se pierde el orden de las palabras. Por ejemplo, "¿Cuándo conviene usar la mediana en lugar de la media?" y "¿Cuándo conviene usar la media en lugar de la mediana?" contienen prácticamente las mismas palabras, pero no expresan la misma pregunta.

Para este último problema podrían utilizarse en el futuro modelos contextuales que permiten representar mejor el contexto y las relaciones entre las palabras.

También podría ampliarse el dataset incorporando diferentes formas de realizar una misma consulta. Otra posibilidad sería establecer un valor mínimo de similitud para que el chatbot pueda indicar que no encontró una respuesta adecuada en lugar de devolver siempre la pregunta más cercana.

En conclusión, el rendimiento de un chatbot de recuperación no depende solamente del método utilizado, sino también del preprocesamiento del texto, la calidad y variedad del dataset y la forma en que los usuarios realizan las consultas.

### Preparación de la base de conocimiento

Se utilizará el dataset original del TP3, compuesto por 40 preguntas y respuestas sobre Estadística Descriptiva, como base de conocimiento del sistema RAG.

Cada par de pregunta y respuesta será tratado como un documento independiente. Luego, estos documentos serán transformados en embeddings y almacenados en una base de datos vectorial ChromaDB.

A cada documento se le asignará además un identificador numérico, que permitirá reconocer posteriormente qué información fue recuperada por el sistema.

```python
# Conversión del dataset original en documentos de LangChain
documentos = []

for i in range(len(preguntas)):
    texto = f"Pregunta: {preguntas[i]}\nRespuesta: {respuestas[i]}"

    documento = Document(
        page_content=texto,
        metadata={"id": i + 1}
    )

    documentos.append(documento)

# Verificación
print("Cantidad de documentos:", len(documentos))
print()
print(documentos[0])
```

### Carga de los modelos de embeddings

Se define una función para cargar los modelos de embeddings seleccionados. Los modelos se ejecutarán en CPU y se utilizará normalización de los vectores para facilitar la comparación por similitud.

```python
def cargar_embedding(nombre_modelo):
    # Se fuerza el uso de CPU
    model_kwargs = {"device": "cpu"}

    # Normalización de los vectores generados
    encode_kwargs = {"normalize_embeddings": True}

    embedding = HuggingFaceEmbeddings(
        model_name=nombre_modelo,
        model_kwargs=model_kwargs,
        encode_kwargs=encode_kwargs
    )

    return embedding
```

### Creación de las bases vectoriales

Se generan dos bases vectoriales en ChromaDB utilizando los mismos documentos, pero diferentes modelos de embeddings. Esto permite comparar posteriormente el comportamiento de ambos modelos bajo las mismas condiciones.

```python
# Modelo de embeddings 1: all-MiniLM-L6-v2
embedding_1 = cargar_embedding("all-MiniLM-L6-v2")

db_1 = Chroma.from_documents(
    documents=documentos,
    embedding=embedding_1,
    collection_name="estadistica_minilm"
)

# Modelo de embeddings 2: sentence_similarity_spanish_es
embedding_2 = cargar_embedding(
    "hiiamsid/sentence_similarity_spanish_es"
)

db_2 = Chroma.from_documents(
    documents=documentos,
    embedding=embedding_2,
    collection_name="estadistica_spanish"
)
```

### Configuración de los recuperadores

A partir de cada base vectorial se crea un recuperador (`retriever`) encargado de buscar los documentos más relacionados con la consulta realizada por el usuario.

En ambos casos se configuró `k = 3`, por lo que el sistema recuperará los tres documentos considerados más similares para cada pregunta.

```python
# Recuperadores: cada uno utiliza una base vectorial diferente.
# k = 3 indica que se recuperarán los tres documentos más similares.
retriever_1 = db_1.as_retriever(
    search_kwargs={"k": 3}
)

retriever_2 = db_2.as_retriever(
    search_kwargs={"k": 3}
)
```

### Carga del modelo de lenguaje

Se carga el modelo **Qwen/Qwen2.5-0.5B-Instruct**, que será utilizado para generar la respuesta final a partir del contexto recuperado.

El `tokenizer` convierte el texto en tokens que pueden ser procesados por el modelo, mientras que `AutoModelForCausalLM` carga el modelo generativo.

```python
# Nombre del modelo de lenguaje seleccionado
modelo_llm = "Qwen/Qwen2.5-0.5B-Instruct"

# Carga del tokenizador, encargado de convertir el texto en tokens
tokenizer = AutoTokenizer.from_pretrained(
    modelo_llm
)

# Carga del modelo generativo
modelo = AutoModelForCausalLM.from_pretrained(
    modelo_llm,
    torch_dtype="auto",
    device_map="auto"
)

print("Modelo cargado correctamente")
```

```python
# Plantilla que recibirá el contexto recuperado y la pregunta del usuario
# {contexto} será reemplazado por los documentos recuperados desde ChromaDB.
# {pregunta} será reemplazado por la consulta realizada por el usuario.
PROMPT_TEMPLATE = """
Respondé la pregunta utilizando únicamente la información del siguiente contexto.

Contexto:
{contexto}

Pregunta:
{pregunta}

Respondé de forma clara y breve en español.
"""
```

```python
class ChatBot:

    def __init__(self, retriever, tokenizer, modelo):
        # Componentes que utilizará cada chatbot
        self.retriever = retriever
        self.tokenizer = tokenizer
        self.modelo = modelo

    def responder(self, pregunta):

        # 1. Recuperar los tres documentos más relevantes
        documentos = self.retriever.invoke(pregunta)

        # 2. Unir el contenido de los documentos recuperados
        # para formar un único contexto
        contexto = "\n\n".join(
            doc.page_content for doc in documentos
        )

        # 3. Completar la plantilla con el contexto
        # y la pregunta realizada por el usuario
        prompt = PROMPT_TEMPLATE.format(
            contexto=contexto,
            pregunta=pregunta
        )

        # 4. Preparar el mensaje con el formato esperado por Qwen
        mensajes = [
            {
                "role": "user",
                "content": prompt
            }
        ]

        texto = self.tokenizer.apply_chat_template(
            mensajes,
            tokenize=False,
            add_generation_prompt=True
        )

        # 5. Convertir el texto en tokens
        entradas = self.tokenizer(
            texto,
            return_tensors="pt"
        ).to(self.modelo.device)

        # 6. Generar la respuesta
        salida = self.modelo.generate(
            **entradas,
            max_new_tokens=150,
            do_sample=False
        )

        # 7. Conservar solamente los tokens nuevos
        # generados por el modelo
        tokens_generados = salida[0][
            entradas["input_ids"].shape[-1]:
        ]

        # 8. Convertir los tokens generados nuevamente en texto
        respuesta = self.tokenizer.decode(
            tokens_generados,
            skip_special_tokens=True
        )

        # También se devuelven los documentos recuperados
        # porque luego serán utilizados para la evaluación
        return respuesta, documentos
```

### Creación de los chatbots

Se crean dos chatbots que utilizan el mismo modelo de lenguaje y el mismo tokenizador. La única diferencia entre ellos es el recuperador asociado a cada modelo de embeddings.

De esta manera, ambos sistemas pueden compararse manteniendo iguales los demás componentes del RAG.

```python
chatbot_1 = ChatBot(
    retriever_1,
    tokenizer,
    modelo
)

chatbot_2 = ChatBot(
    retriever_2,
    tokenizer,
    modelo
)
```

## d) Evaluación y comparación de los modelos de embeddings

Una vez implementados los dos chatbots, se realizará una evaluación utilizando el conjunto de 15 preguntas de prueba definido previamente.

Ambos chatbots utilizarán la misma base de conocimiento, el mismo modelo de lenguaje y la misma cantidad de documentos recuperados. La única diferencia entre ellos será el modelo de embeddings utilizado:

- **Chatbot 1:** `all-MiniLM-L6-v2`
- **Chatbot 2:** `hiiamsid/sentence_similarity_spanish_es`

Para cada pregunta se almacenará:

- la pregunta realizada;
- la respuesta generada por el modelo;
- los contextos recuperados desde la base vectorial.

Esto permitirá comparar el comportamiento de ambos modelos de embeddings y analizar cuál recupera información más relevante para responder correctamente.

### Ejecución de las preguntas de evaluación

Se define una función que recorre las 15 preguntas de prueba y utiliza cada chatbot para generar una respuesta.

Además de la respuesta generada, se almacenan los documentos recuperados, ya que posteriormente serán utilizados para analizar el comportamiento de cada modelo de embeddings.

```python
def evaluar_chatbot(chatbot, preguntas):
    resultados = []

    # Recorre todas las preguntas de evaluación
    for i, pregunta in enumerate(preguntas, start=1):

        # Muestra el avance del proceso
        print(
            f"Procesando pregunta {i} de {len(preguntas)}..."
        )

        # El chatbot devuelve la respuesta generada
        # y los documentos recuperados
        respuesta, documentos = chatbot.responder(
            pregunta
        )

        # Se guardan los resultados de cada pregunta
        resultados.append({
            "pregunta": pregunta,
            "respuesta": respuesta,
            "contextos": [
                doc.page_content
                for doc in documentos
            ]
        })

    return resultados
```

### Evaluación técnica de la recuperación

Además de observar las respuestas generadas, se analiza técnicamente la recuperación realizada por cada modelo de embeddings.

Para cada pregunta se registran:

- el ID del documento recuperado en primer lugar;
- los IDs de los tres documentos recuperados;
- la similitud entre la consulta y el primer documento recuperado;
- el tiempo de búsqueda;
- la respuesta generada por el chatbot.

Esto permite comparar no solo las respuestas finales, sino también el comportamiento de cada modelo durante la etapa de recuperación.

```python
import time
import numpy as np
```

```python
def evaluar_recuperacion(db,embedding,preguntas,resultados_chatbot,nombre_modelo):
    filas = []

    # Se evalúa cada pregunta de prueba
    for i, pregunta in enumerate(preguntas):

        # Medición del tiempo de búsqueda
        inicio = time.time()

        documentos = db.similarity_search(
            pregunta,
            k=3
        )

        tiempo_ms = (time.time() - inicio) * 1000

        # Se obtiene el vector de la pregunta
        vector_pregunta = np.array(
            embedding.embed_query(pregunta)
        )

        # Se obtiene el vector del documento recuperado en primer lugar
        vector_top1 = np.array(
            embedding.embed_query(
                documentos[0].page_content
            )
        )

        # Cálculo de similitud coseno entre ambos vectores
        similitud = np.dot(
            vector_pregunta,
            vector_top1
        ) / (
            np.linalg.norm(vector_pregunta)
            * np.linalg.norm(vector_top1)
        )

        # Se guardan los resultados de la recuperación
        filas.append({
            "Modelo": nombre_modelo,
            "Pregunta": pregunta,
            "Ground truth": ground_truth[i],
            "ID Top 1": documentos[0].metadata["id"],
            "IDs Top 3": [
                doc.metadata["id"]
                for doc in documentos
            ],
            "Similitud Top 1": round(
                float(similitud),
                3
            ),
            "Tiempo búsqueda (ms)": round(
                tiempo_ms,
                2
            ),
            "Respuesta generada":
                resultados_chatbot[i]["respuesta"]
        })

    return pd.DataFrame(filas)
```

```python
# Evaluación técnica del sistema que utiliza MiniLM
tabla_minilm = evaluar_recuperacion(
    db_1,
    embedding_1,
    preguntas_prueba,
    resultados_1,
    "MiniLM"
)

# Evaluación técnica del sistema que utiliza SentenceSimilarityES
tabla_sentence_similarity = evaluar_recuperacion(
    db_2,
    embedding_2,
    preguntas_prueba,
    resultados_2,
    "SentenceSimilarityES"
)
```

```python
# Se combinan los resultados técnicos de ambos modelos
# en una única tabla para facilitar la comparación.
comparacion_tecnica = pd.concat(
    [tabla_minilm, tabla_sentence_similarity],
    ignore_index=True
)

pd.set_option("display.max_colwidth", None)

display(comparacion_tecnica)
```

## e) Evaluación de los modelos de embeddings

Como evaluación adicional, se analizarán los modelos de embeddings mediante las métricas **Context Precision** y **Context Recall**.

- **Context Precision** permite evaluar qué tan relevante es la información recuperada para responder una consulta.
- **Context Recall** permite analizar si los contextos recuperados contienen la información necesaria para responder de acuerdo con la respuesta esperada (`ground_truth`).

Se evaluarán ambos modelos utilizando las mismas 15 preguntas, con el objetivo de comprobar si los resultados obtenidos respaldan la elección realizada en el punto anterior.

### Organización de los contextos recuperados

### Definición de documentos relevantes

Para calcular las métricas de recuperación se define, para cada pregunta de evaluación, qué documentos de la base de conocimiento se consideran relevantes.

En algunos casos existe más de un documento válido, ya que diferentes preguntas de la base pueden aportar información útil para responder una misma consulta.

### Cálculo de Context Precision y Context Recall

Se define una función que compara los documentos recuperados por cada modelo con los documentos considerados relevantes para cada pregunta.

- **Context Recall** mide qué proporción de los documentos relevantes fue recuperada.
- **Context Precision** considera además la posición en la que aparecen los documentos relevantes, dando mayor valor a aquellos que se recuperan en los primeros lugares.

De esta manera, no solo se analiza si el sistema encontró información correcta, sino también qué tan bien ordenada aparece dentro de los resultados recuperados.

```python
def calcular_metricas_ids(ids_recuperados, ids_esperados):

    # Se marca con 1 si el documento recuperado es relevante
    # y con 0 si no pertenece a los documentos esperados.
    relevancias = [
        1 if id_doc in ids_esperados else 0
        for id_doc in ids_recuperados
    ]

    # -------- Context Recall --------
    # Proporción de documentos relevantes que fueron recuperados.
    recuperados_correctos = len(
        set(ids_recuperados) & set(ids_esperados)
    )

    context_recall = (
        recuperados_correctos / len(ids_esperados)
        if ids_esperados
        else 0
    )

    # -------- Context Precision --------
    # Evalúa si los documentos relevantes aparecen
    # en las primeras posiciones de la recuperación.
    precisiones = []
    relevantes_acumulados = 0

    for posicion, relevante in enumerate(
        relevancias,
        start=1
    ):
        if relevante == 1:
            relevantes_acumulados += 1
            precisiones.append(
                relevantes_acumulados / posicion
            )

    context_precision = (
        sum(precisiones) / len(precisiones)
        if precisiones
        else 0
    )

    return context_precision, context_recall
```

### Aplicación de las métricas a ambos modelos

Se aplican las métricas anteriores a los resultados obtenidos por `MiniLM` y `SentenceSimilarityES`.

Para cada pregunta se comparan los IDs recuperados con los IDs esperados y se registran los valores de `Context Precision` y `Context Recall`.

```python
filas_metricas = []

# Se recorren los resultados técnicos de ambos modelos.
for _, fila in comparacion_tecnica.iterrows():

    pregunta = fila["Pregunta"]
    ids_recuperados = fila["IDs Top 3"]

    # Documentos considerados relevantes para esa pregunta
    ids_correctos = ids_esperados[pregunta]

    # Cálculo de las dos métricas
    precision, recall = calcular_metricas_ids(
        ids_recuperados,
        ids_correctos
    )

    # Se almacenan los resultados de cada pregunta
    filas_metricas.append({
        "Modelo": fila["Modelo"],
        "Pregunta": pregunta,
        "ID esperado": ids_correctos,
        "IDs recuperados": ids_recuperados,
        "Context Precision": round(precision, 3),
        "Context Recall": round(recall, 3)
    })

metricas_ids_df = pd.DataFrame(filas_metricas)

display(metricas_ids_df)
```

### Resultados promedio por modelo

Finalmente, se calcula el promedio de `Context Precision` y `Context Recall` para cada modelo de embeddings.

Esto permite obtener una visión general del desempeño de cada modelo en la recuperación de información relevante.

```python
# Se agrupan los resultados por modelo
# y se calculan los valores promedio de ambas métricas.
resumen_metricas = metricas_ids_df.groupby("Modelo").agg({
    "Context Precision": "mean",
    "Context Recall": "mean"
}).reset_index()

# Redondeo para facilitar la lectura
resumen_metricas["Context Precision"] = (
    resumen_metricas["Context Precision"].round(3)
)

resumen_metricas["Context Recall"] = (
    resumen_metricas["Context Recall"].round(3)
)

display(resumen_metricas)
```

- **SentenceSimilarityES:** Context Precision ≈ 1.000 y Context Recall ≈ 0.778.

Es importante señalar que **Context Precision** y **Context Recall** evalúan la etapa de recuperación del sistema RAG, es decir, qué tan adecuados son los documentos encontrados por el modelo de embeddings. No evalúan directamente la calidad de la respuesta generada por el modelo de lenguaje.

Los resultados también permiten observar que la calidad de la recuperación y la calidad de la respuesta final no siempre coinciden. Por ejemplo, ante algunas consultas un modelo puede recuperar documentos más adecuados y, aun así, el modelo de lenguaje puede generar una respuesta menos precisa.

En conclusión, `all-MiniLM-L6-v2` mostró ventajas en velocidad y un mejor Context Recall, mientras que `SentenceSimilarityES` obtuvo un mejor Context Precision.
