##  2) Crear el chatbot utilizando TFIDF y similitud del coseno. (1 punto)

**Uso de IA:** Se utilizó ChatGPT como apoyo para comprender el funcionamiento
de `TfidfVectorizer`, la similitud del coseno y para revisar la estructura de
la función de recuperación. El código fue probado y analizado durante el
desarrollo del trabajo.

### Preguntas de Prueba

```python
preguntas_prueba = [

    # 5 consultas con vocabulario parecido al dataset
    "¿Qué significa población en estadística?",
    "¿Cómo se obtiene la frecuencia relativa?",
    "¿Cómo puedo calcular la media aritmética?",
    "¿Qué significa desviación estándar?",
    "¿Para qué se utiliza un gráfico de barras?",

    # 10 reformulaciones naturales
    "¿Qué llamamos muestra en un estudio estadístico?",
    "¿Cómo puedo saber si una variable es cualitativa o cuantitativa?",
    "¿En qué se distingue una variable discreta de una continua?",
    "¿Qué indica la frecuencia absoluta de un dato?",
    "¿Para qué sirve organizar los datos en una tabla de frecuencias?",
    "¿Cómo saco el promedio de un conjunto de valores?",
    "¿Qué representa la mediana de un conjunto de datos?",
    "¿Cómo sé cuál es la moda de un conjunto de datos?",
    "¿En qué casos conviene usar la mediana y no la media?",
    "¿Qué diferencia hay entre un histograma y un gráfico de barras?"
]
```

```python
# Crear el vectorizador
vectorizer = TfidfVectorizer(
    lowercase=True,
    strip_accents="unicode"
)

# Ajustar y transformar las preguntas del dataset
tfidf_matrix = vectorizer.fit_transform(preguntas)

# Función para encontrar la mejor coincidencia
def responder_pregunta(consulta_usuario):

    # Transformar la consulta usando el mismo vectorizador
    consulta_vec = vectorizer.transform([consulta_usuario])

    # Calcular la similitud entre la consulta y todas las preguntas
    similitudes = cosine_similarity(
        consulta_vec,
        tfidf_matrix
    ).flatten()

    # Buscar el índice de la pregunta con mayor similitud
    indice_mejor_coincidencia = similitudes.argmax()

    # Guardar los resultados
    respuesta = respuestas[indice_mejor_coincidencia]
    pregunta_encontrada = preguntas[indice_mejor_coincidencia]
    similitud = similitudes[indice_mejor_coincidencia]

    return respuesta, pregunta_encontrada, similitud
```

```python
for consulta in preguntas_prueba:

    respuesta, pregunta_encontrada, similitud = responder_pregunta(consulta)

    print("Consulta:", consulta)
    print("Pregunta encontrada:", pregunta_encontrada)
    print("Similitud:", round(similitud, 3))
    print("Respuesta:", respuesta)
    print("-" * 100)
```

En las 15 consultas de prueba, el chatbot basado en TF-IDF obtuvo 9 respuestas correctas, 4 parcialmente correctas y 2 incorrectas.

El modelo mostró un buen desempeño cuando las consultas mantenían términos importantes presentes en las preguntas del dataset. Sin embargo, presentó dificultades ante algunas reformulaciones que utilizaban palabras diferentes para expresar un mismo concepto. Por ejemplo, la consulta "¿Cómo saco el promedio de un conjunto de valores?" no fue relacionada con "media aritmética".

```python
vectores_preguntas = []

for pregunta in preguntas:
    vector = nlp(pregunta).vector
    vectores_preguntas.append(vector)
```

```python
vectores_preguntas = np.array(vectores_preguntas)
vectores_preguntas.shape
```

```python
def responder_embeddings(consulta_usuario):

    # Convertir la consulta del usuario en un embedding
    consulta_vec = nlp(consulta_usuario).vector

    # Comparar la consulta con los vectores de las 40 preguntas
    similitudes = cosine_similarity(
        [consulta_vec],
        vectores_preguntas
    ).flatten()

    # Buscar el índice de la mayor similitud
    indice_mejor_coincidencia = similitudes.argmax()

    # Recuperar la pregunta, respuesta y similitud
    respuesta = respuestas[indice_mejor_coincidencia]
    pregunta_encontrada = preguntas[indice_mejor_coincidencia]
    similitud = similitudes[indice_mejor_coincidencia]

    return respuesta, pregunta_encontrada, similitud
```

```python
for consulta in preguntas_prueba:

    respuesta, pregunta_encontrada, similitud = responder_embeddings(consulta)

    print("Consulta:", consulta)
    print("Pregunta encontrada:", pregunta_encontrada)
    print("Similitud:", round(similitud, 3))
    print("Respuesta:", respuesta)
    print("-" * 80)
```

```python
def obtener_vector_sin_stopwords(texto):

    doc = nlp(texto)

    vectores = [
        token.vector
        for token in doc
        if not token.is_stop
        and not token.is_punct
        and not token.is_space
        and token.has_vector
    ]

    if len(vectores) > 0:
        return np.mean(vectores, axis=0)

    return np.zeros(nlp.vocab.vectors_length)
```

```python
vectores_preguntas_sin_stopwords = []

for pregunta in preguntas:
    vector = obtener_vector_sin_stopwords(pregunta)
    vectores_preguntas_sin_stopwords.append(vector)

vectores_preguntas_sin_stopwords = np.array(
    vectores_preguntas_sin_stopwords
)
```

```python
def responder_embeddings_sin_stopwords(consulta_usuario):

    # Convertir la consulta usando el mismo procesamiento
    consulta_vec = obtener_vector_sin_stopwords(consulta_usuario)

    # Comparar con las 40 preguntas
    similitudes = cosine_similarity(
        [consulta_vec],
        vectores_preguntas_sin_stopwords
    ).flatten()

    # Buscar la mayor similitud
    indice_mejor_coincidencia = similitudes.argmax()

    # Recuperar resultados
    respuesta = respuestas[indice_mejor_coincidencia]
    pregunta_encontrada = preguntas[indice_mejor_coincidencia]
    similitud = similitudes[indice_mejor_coincidencia]

    return respuesta, pregunta_encontrada, similitud
```

```python
for consulta in preguntas_prueba:

    respuesta, pregunta_encontrada, similitud = responder_embeddings_sin_stopwords(
        consulta
    )

    print("Consulta:", consulta)
    print("Pregunta encontrada:", pregunta_encontrada)
    print("Similitud:", round(similitud, 3))
    print("Respuesta:", respuesta)
    print("-" * 80)
```

## 5. Conclusiones

A partir de las pruebas realizadas, ambos chatbots lograron responder correctamente varias consultas, aunque tuvieron dificultades diferentes.

El modelo **TF-IDF** funcionó mejor cuando las consultas mantenían palabras similares a las del dataset, pero presentó dificultades cuando se utilizaron otras formas de expresar una misma idea.

En el caso de **spaCy con `es_core_news_md`**, la primera implementación tuvo un rendimiento bajo. Al eliminar stopwords, signos de puntuación y espacios, los resultados mejoraron considerablemente, pasando de 4 a 9 respuestas correctas. También se probó el modelo `es_core_news_lg`, pero no presentó una mejora importante, por lo que se mantuvo el modelo `md`.

En general, ninguno de los dos métodos fue claramente superior. TF-IDF tuvo menos respuestas incorrectas, mientras que spaCy logró reconocer mejor algunas reformulaciones. También se comprobó que una similitud alta no significa necesariamente que la respuesta sea correcta.

### Posibles mejoras

En **TF-IDF** se podría incorporar lematización para tratar de la misma manera distintas formas de una palabra. Por ejemplo, "calcula", "calcular" y "calculando" podrían reducirse al lema "calcular". También podrían utilizarse bigramas para considerar combinaciones de palabras y no solamente como términos separados.

En **spaCy**, la eliminación de stopwords puede hacer perder información importante. Por ejemplo, eliminar la palabra "no" podría modificar completamente la intención de una consulta. Además, en esta implementación el vector de cada pregunta se obtuvo calculando el promedio de los embeddings de sus palabras. Al utilizar un promedio se pierde el orden de las palabras. Por ejemplo, "¿Cuándo conviene usar la mediana en lugar de la media?" y "¿Cuándo conviene usar la media en lugar de la mediana?" contienen prácticamente las mismas palabras, pero no expresan la misma pregunta.

Para este último problema podrían utilizarse en el futuro modelos contextuales que permiten representar mejor el contexto y las relaciones entre las palabras.

También podría ampliarse el dataset incorporando diferentes formas de realizar una misma consulta. Otra posibilidad sería establecer un valor mínimo de similitud para que el chatbot pueda indicar que no encontró una respuesta adecuada en lugar de devolver siempre la pregunta más cercana.

En conclusión, el rendimiento de un chatbot de recuperación no depende solamente del método utilizado, sino también del preprocesamiento del texto, la calidad y variedad del dataset y la forma en que los usuarios realizan las consultas.
